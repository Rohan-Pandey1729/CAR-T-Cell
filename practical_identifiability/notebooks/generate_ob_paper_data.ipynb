```python
def plot_TC_with(params, init_vals, t_max):
    gt1 = owens_bozic_model.get_ground_truth(params, init_vals, 0, t_max)
    t = np.linspace(0, t_max, 100)
    obs = gt1(t)
    for obs_var in ["T", "C"]:
        plt.semilogy(t, np.maximum(obs[obs_var], 1), label=obs_var)
    plt.legend()
    plt.show()
```

```python
plot_TC_with(
    p1_params,
    {
        "T": 5e9,
        "E": 4e5,
        "C": 1e7,
        "M": 6,
    },
    150
)
```
