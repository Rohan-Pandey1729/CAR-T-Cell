```python
import pandas as pd
import os
```

```python
# join csvs
PATH = "./raw_data"
FIGURES = ["2B", "3A", "3B"]

fig_to_df: dict[str, pd.DataFrame] = {}
for fp in os.listdir(PATH):
    for fig in FIGURES:
        if fp.startswith(fig):
            fp = os.path.join(PATH, fp)
            if fig not in fig_to_df:
                fig_to_df[fig] = pd.read_csv(fp, header=None)
            else:
                df = pd.read_csv(fp, header=None)
                fig_to_df[fig] = pd.concat([fig_to_df[fig], df], axis=0, ignore_index=True)
```

```python
# set column names
for fig in FIGURES:
    df = fig_to_df[fig]
    df.columns = ["day", "y", "_ignore", "patient"]
    del df["_ignore"]
```

```python
# save combined dfs
for fig in FIGURES:
    fig_to_df[fig].to_csv(f"{fig}.csv", index=False)
```
